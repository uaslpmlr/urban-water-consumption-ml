```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

# Detect the repository root
current_dir = Path.cwd()

if (current_dir / "data").exists():
    repository_root = current_dir
elif (current_dir.parent / "data").exists():
    repository_root = current_dir.parent
else:
    raise FileNotFoundError(
        "The repository data directory could not be found. "
        "Run the notebook from the repository root or from the notebooks directory."
    )

# Dataset path
data_path = repository_root / "data" / "Ini23NovFin30Abril.csv"

# Verify that the dataset exists
if not data_path.exists():
    raise FileNotFoundError(
        "Dataset not found. Place 'Ini23NovFin30Abril.csv' "
        "inside the repository data directory."
    )

# Load and prepare the dataset
df = pd.read_csv(data_path)
df = df.loc[:, ~df.columns.str.contains(r"^Unnamed")]
df["Fecha"] = pd.to_datetime(df["Fecha"], dayfirst=True)

# Verify the required column
if "Consumo" not in df.columns:
    raise KeyError(
        "The dataset must contain a column named 'Consumo'."
    )

# Remove missing consumption values
consumption = df["Consumo"].dropna()

# Create the histogram
fig, ax = plt.subplots(figsize=(8, 5))

sns.histplot(
    consumption,
    bins=20,
    kde=True,
    color="#0072B2",
    edgecolor="black",
    alpha=1,
    ax=ax
)

# Format the kernel density estimation curve
for line in ax.lines:
    line.set_color("black")
    line.set_linewidth(2)

# Axis labels and formatting
ax.set_xlabel("Daily Consumption (m³)", fontsize=16)
ax.set_ylabel("Frequency", fontsize=16)
ax.tick_params(axis="both", labelsize=14)
ax.grid(True, linestyle="--", alpha=0.4)

fig.tight_layout()
plt.show()
```
